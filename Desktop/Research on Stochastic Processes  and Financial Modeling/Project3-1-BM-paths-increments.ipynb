```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define a function which takes as input the length t of the interval [0,t] and the step size t/n, and has output the Brownian
# path (B_0, B_{t1}, B_{t2},... B_{t})

def brownian(interval, step):    
    n = int(interval/step)        # computes the integer value
    B = [0]
    for i in range(n):
        B.append(B[-1] + np.sqrt(step) * np.random.standard_normal())
    return B
```

```python
# Computes the Brownian path for t=1 and n=100

B = brownian(1, 0.01)
```

```python
# Plot one Brownian path

x = np.linspace(0.0, 1.0, 101)      #creates the array 0,0.01,0.02,...,0.99,1
plt.plot(x, B)
```

```python
# Plot 5 Brownian paths

B = []
for i in range(5):
  B.append(brownian(1, 0.01))
  plt.plot(x, B[i])
```
