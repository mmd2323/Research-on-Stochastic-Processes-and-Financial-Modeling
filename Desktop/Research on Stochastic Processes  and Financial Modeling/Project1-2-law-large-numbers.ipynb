```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
# Plot S_N/N for N=1,...,10000 where S_N=X_1+...+X_N and X1,...,X_N is a sample from Expo(1)
# Plot should be approximately 1 (SLLN)

x = np.random.exponential(1,10000)
s = np.cumsum(x)                    # computes cumulative sums of x
l = np.arange(1,10001)            # produces the array [1,2,...,10000]
r = s/l
plt.plot(l,r)

```

```python
# Generate M=8000 samples X_1,...,X_N of size N=10000 from Expo(1) and compute Y=S_N/N for each sample
# Plot cdf histogram of the 8000 values Y: histogram approximates the function F(x)=0 if x<1, F(x)=1 if x>=1 (WLLN)

def emp_mean(a,N):
    return np.cumsum(a,axis=0)[-1]/N      #gives the last row of NxM array which contains the cum-sums over rows divided by N

x = np.random.exponential(1,(10000,8000))
y = emp_mean(x,10000)
n, bins, patches = plt.hist(y, 50, density=True, cumulative = True)
```
