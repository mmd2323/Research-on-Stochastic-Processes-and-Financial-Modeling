```python
import numpy as np
import matplotlib.pyplot as plt
```

**Compute Ito integral $\int_0^t B_s dB_s$**

```python
# Plot one path of the integral \int_0^t B_s dB_s

x = np.linspace(0.0, 1.0, 1001)      #creates the array 0,0.001,0.002,...,0.999,1

B = [0]
X = [0]
I = [0]
for j in range(1000):
    B.append(B[-1] + np.sqrt(0.001) * np.random.standard_normal())
    X.append(B[-1])
    I.append(I[-1] + X[-2] * (B[-1] - B[-2]))
plt.plot(x, I)   
```

```python
# Plot 5 paths of the Ito integral \int_0^t B_s dB_s
x = np.linspace(0.0, 1.0, 1001)

P = []
for i in range(5):
    B = [0]
    X = [0]
    I = [0]
    for j in range(1000):
        B.append(B[-1] + np.sqrt(0.001) * np.random.standard_normal())
        X.append(B[-1])
        I.append(I[-1] + X[-2] * (B[-1] - B[-2]))
    P.append(I)
    plt.plot(x, P[i])
```

**Compute Ito integral $\int_0^t 4B_s^3 dB_s$**

```python
# Plot one path of the integral \int_0^t 4B_s^3 dB_s

x = np.linspace(0.0, 1.0, 1001)      #creates the array 0,0.001,0.002,...,0.999,1
B = [0]
X1 = [0]
I1 = [0]
for i in range(1000):
    B.append(B[-1] + np.sqrt(0.001) * np.random.standard_normal())
    X1.append(4 * B[-1]**3)
    I1.append(I1[-1] + X1[-2] * (B[-1] - B[-2]))
plt.plot(x, I1)   
```

```python
# Plot 5 paths of the Ito integral \int_0^t 4B_s^3 dB_s
x = np.linspace(0.0, 1.0, 1001)

P1 = []
for i in range(5):
    B = [0]
    X1 = [0]
    I1 = [0]
    for j in range(1000):
        B.append(B[-1] + np.sqrt(0.001) * np.random.standard_normal())
        X1.append(4 * B[-1]**3)
        I1.append(I1[-1] + X1[-2] * (B[-1] - B[-2]))
    P1.append(I1)
    plt.plot(x, P1[i])
```

**Compute Ito integral $\int_0^t \cos(B_s)dB_s$**

```python
# Plot one path of the integral \int_0^t \cos(B_s) dB_s

x = np.linspace(0.0, 1.0, 1001)      #creates the array 0,0.001,0.002,...,0.999,1

B = [0]
X2 = [0]
I2 = [0]
for j in range(1000):
    B.append(np.sqrt(0.001) * np.random.standard_normal() + B[-1])
    X2.append(np.cos(B[-1]))
    I2.append(I2[-1] + X2[-2] * (B[-1] - B[-2]))
plt.plot(x, I2)       
```

```python
# Plot 5 paths of the integral \int_0^t \cos(B_s) dB_s

P2 = []
for i in range(5):
  B = [0]
  X2 = [0]
  I2 = [0]
  for j in range(1000):
    B.append(np.sqrt(0.001) * np.random.standard_normal() + B[-1])
    X2.append(np.cos(B[-1]))
    I2.append(I2[-1] + X2[-2] * (B[-1] - B[-2]))
  P2.append(I2)
  plt.plot(P2[i])
```
