```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define a function which takes as input the length t of the interval [0,t] and the step size t/n, and has output the Brownian
# path (B_0, B_{t1}, B_{t2},... B_{t})

def brownian(interval, step):    
    n = int(interval/step)        # computes the integer value
    B = [0]
    for i in range(n):
        B.append(B[-1] + np.sqrt(step) * np.random.standard_normal())
    return B
```

```python
# Computes the Brownian path for t=1 and step 2^{-20}

B = brownian(1, 1/2**20)
```

```python
# Define a function which computes the variation of Brownian path

def variation(n):
  Bn = np.array(B[::2**(20-n)])    # extract points from BP in steps on 2^{20-n}
  B1 = Bn[1:]                      # remove the first element of Bn
  B2 = Bn[:-1]                     # remove the last element of Bn
  V = np.abs(B1 - B2)              # compute absolute value of increments
  return np.sum(V)                 # return sum of absolute value of increments
```

```python
x = np.linspace(10, 20, 11)
V = []
for i in range(10,21):
  Vi = variation(i)
  print(Vi)          
  V.append(Vi)
plt.plot(x, V)        #plot (i,variation(i)) for i=10,11,...,20
```

```python
#Defines a function which computes the quadratic variation of Brownian path

def quad_variation(n):
  Bn = np.array(B[::2**(20-n)])      # extract points from BP in steps on 2^{20-n}
  B1 = Bn[1:]                        # remove the first element of Bn
  B2 = Bn[:-1]                       # remove the last element of Bn
  W = np.square(B1 - B2)             # compute squared increments
  return np.sum(W)                   # return sum of squared increments
```

```python
x = np.linspace(10, 20, 11)
Q = []
for i in range(10,21):
  Qi = quad_variation(i)
  print(Qi)
  Q.append(Qi)
plt.plot(x, Q)      #plot (i,quad_variation(i)) for i=10,11,...,20
```
