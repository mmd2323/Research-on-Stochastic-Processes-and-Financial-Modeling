```python
import numpy as np
import random
import matplotlib.pyplot as plt

```

```python
C = np.zeros((100,100))
for i in range(100):
    for j in range(100):
        C[i][j]=min(i+1,j+1)/100
A = np.linalg.cholesky(C)
Z = np.random.standard_normal((100,5))
X = np.dot(A,Z)
Zero = np.zeros((1,5))
Y = np.vstack([Zero,X])
B = np.transpose(Y)

```

```python
# Plot of 5 sample paths of Brownian motion

x=np.linspace(0,1,101)
for i in range(5):
    plt.plot(x,B[i])
```
