```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
# density histogram of a sample X_1,...,X_N from Uniform(0,1) with N=10000 (50 bins)

data1 = np.random.rand(10000)
n, bins, patches = plt.hist(data1, 50, density=True)
```

```python
# cumulative density histogram of a sample X_1,...,X_N from Uniform(0,1) with N=10000 (50 bins)

n, bins, patches = plt.hist(data1, 50, density=True, cumulative=True)
```

```python
# density histogram of a sample X_1,...,X_N from N(2,25) with N=10000 (50 bins)

data2=np.random.normal(2,5,10000)
n, bins, patches = plt.hist(data2, 50, density=True)
```
