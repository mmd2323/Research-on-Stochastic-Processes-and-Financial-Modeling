```python
import numpy as np
import matplotlib.pyplot as plt
```

**Compute Ito integral $\int_0^t B_s dB_s$**

```python
# Define a function which takes as input the length T of the interval [0,T] and the step size T/n, and has output the path
# (I_0, I_{t1}, I_{t2},... I_{T}) of the Ito integral I_t =\int_0^t B_s dB_s

def integral(interval, step):    
    n = int(interval/step)
    B = [0]
    X = [0]
    I = [0]
    for i in range(n):
        B.append(B[-1] + np.sqrt(step) * np.random.standard_normal())
        X.append(B[-1])
        I.append(I[-1] + X[-2] * (B[-1] - B[-2]))
    return I
```

```python
# Computes one path of the integral \int_0^t B_s dB_s for T=1 and n=1000

I = integral(1, 0.001)
```

```python
# Plot one path of the integral \int_0^t B_s dB_s

x = np.linspace(0.0, 1.0, 1001)      #creates the array 0,0.001,0.002,...,0.999,1
plt.plot(x, I)
```

```python
# Plot 5 paths of the Ito integral \int_0^t B_s dB_s

I = []
for i in range(5):
  I.append(integral(1, 0.001))
  plt.plot(x, I[i])
```

**Compute Ito integral $\int_0^t 4B_s^3 dB_s$**

```python
# Define a function which takes as input the length T of the interval [0,T] and the step size T/n, and has output the path
# (I_0, I_{t1}, I_{t2},... I_{T}) of the Ito integral I_t =\int_0^t 4B_s^3 dB_s

def integral1(interval, step):    
    n = int(interval/step)
    B = [0]
    X1 = [0]
    I1 = [0]
    for i in range(n):
        B.append(B[-1] + np.sqrt(step) * np.random.standard_normal())
        X1.append(4 * B[-1]**3)
        I1.append(I1[-1] + X1[-2] * (B[-1] - B[-2]))
    return I1
```

```python
# Computes one path of the integral \int_0^t 4B_s^3 dB_s for T=1 and n=1000

I1 = integral1(1, 0.001)
```

```python
# Plot one path of the integral \int_0^t 4B_s^3 dB_s

x = np.linspace(0.0, 1.0, 1001)      #creates the array 0,0.001,0.002,...,0.999,1
plt.plot(x, I1)
```
