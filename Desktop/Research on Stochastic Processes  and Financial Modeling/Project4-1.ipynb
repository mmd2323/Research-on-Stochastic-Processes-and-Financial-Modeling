```python
import numpy as np
import matplotlib.pyplot as plt
    
     
```

```python
# Define a function for simulating the compensated squared BM : B_t^2 -t

def b_sqr_compensated(interval, step):
  n = int(interval / step)
  B = [0]
  P = [0]
  for i in range(n):
    t = (i+1) * step
    B.append(np.sqrt(step) * np.random.standard_normal() + B[-1])
    P.append(B[-1]**2 - t)
  return P
     
```

```python
P = []
for i in range(5):
  P.append(b_sqr_compensated(1, 0.01))
```

```python
# Simulate 5 paths pf the compensated squared BM on the interval [0,1] with step size 0.01

x = np.linspace(0.0, 1.0, 101)
for i in range(5):
  plt.plot(x, P[i])
```

```python
# Define a function for simulating the geometric BM : exp(B_t-t/2)

def geo_brownian(interval, step):
  n = int(interval / step)
  BP = [0]
  G = [1]
  for i in range(n):
    t = (i+1) * step
    BP.append(np.sqrt(step) * np.random.standard_normal() + BP[-1])
    G.append(np.exp(BP[-1] - t/2))
  return G
```

```python
G = []
for i in range(5):
  G.append(geo_brownian(1, 0.01))
```

```python
# Simulate 5 paths pf the geometric BM on the interval [0,1] with step size 0.01

x = np.linspace(0.0, 1.0, 101)
for i in range(5):
  plt.plot(x, G[i])
```

```python
# Define a function for simulating the Poisson process  

def poisson(interval, step):
  n = int(interval / step)
  N = [0]
  for i in range(n):
    N.append(np.random.poisson(1 * step) + N[-1])
  return N
```

```python
N = []
for i in range(5):
  N.append(poisson(10, 0.1))
```

```python
# Simulate 5 paths of the Poisson process with rate 1 on the interval [0,10] with step size 0.01

x = np.linspace(0.0, 10.0, 101)
for i in range(5):
  plt.plot(x, N[i])
```

```python
# Simulate 5 paths of the compensated Poisson process with rate 1 on the interval [0,10] with step size 0.01

def comp_poisson(interval, step):
  n = int(interval / step)
  N = [0]
  M = [0]
  for i in range(n):
    t = (i+1) * step
    N.append(np.random.poisson(1 * step) + N[-1])
    M.append(N[-1] - t)
  return M
```

```python
M = []
for i in range(5):
  M.append(comp_poisson(10, 0.1))
```

```python
# Simulate 5 paths of the compensated Poisson process with rate 1 on the interval [0,10] with step size 0.01

x = np.linspace(0.0, 10.0, 101)
for i in range(5):
  plt.plot(x, M[i])
```
