```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
# Generate M=8000 samples X_1,...,X_N of size N=10000 from Expo(1) and compute Y=(S_N-N)/sqrt(N) for each sample
# Plot pdf histogram of the 8000 values Y: histogram approximates N(0,1) pdf (CLT)

def normalized_sum(a,N):
    return (np.cumsum(a, axis=0)[-1] - N) / np.sqrt(N)

x = np.random.exponential(1,(10000,8000))
y = normalized_sum(x,10000)
n, bins, patches = plt.hist(y, 50, density=True)
```

```python
# Generate M=8000 samples X_1,...,X_N of size N=6 from N(2,25) and compute Y=(S_N-2*N)/(5*sqrt(N)) for each sample
# Plot pdf histogram of the 8000 values Y: histogram approximates N(0,1) pdf (standardization, S_N is normal)

def normalized_sum1(a,N):
    return (np.cumsum(a, axis=0)[-1] - 2*N) / (5*np.sqrt(N))

x = np.random.normal(2,5,(6,8000))
y = normalized_sum1(x,6)
n, bins, patches = plt.hist(y, 50, density=True)
```
